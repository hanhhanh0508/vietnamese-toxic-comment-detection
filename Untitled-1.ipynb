```python
import pandas as pd

# Đường dẫn đến file train.csv bạn vừa tải về (chỉnh lại đường dẫn nếu cần)
file_path = "D:/CNTT kì 1-Năm 4/NLP/train.csv"

try:
    # Đọc file CSV với bảng mã UTF-8
    df = pd.read_csv(file_path, encoding="utf-8")
    
    print("=== TẢI DỮ LIỆU THÀNH CÔNG ===")
    print(f"Tổng số dòng: {len(df)}")
    print("\n=== 5 DÒNG ĐẦU TIÊN (TIẾNG VIỆT CHUẨN) ===")
    
    # In ra 5 dòng đầu tiên
    for idx, row in df.head(20).iterrows():
        print(f"[{idx+1}] Text: {row['free_text']} | Label: {row['label_id']}")

except Exception as e:
    print("Có lỗi xảy ra:", e)
```
